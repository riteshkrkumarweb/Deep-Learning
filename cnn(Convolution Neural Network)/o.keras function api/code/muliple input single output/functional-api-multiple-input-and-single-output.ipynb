```python
from keras.models import Model
from keras.layers import *
```

```python
# Here we Define 2 inputs 
input_1 = Input(shape=(64,))
input_2 = Input(shape=(128,))
```

```python
# Here we define the input_1 
input_for_1 = Dense(8,activation='relu')(input_1) # this is the input layer 
hidden_1 = Dense(4,activation='relu')(input_for_1)
```

```python
# Here we define for the input_2 
input_for_2 = Dense(64,activation='relu')(input_2) # this is the input layer 
hidden_2 = Dense(8,activation='relu')(input_for_2)
hidden_2_1 = Dense (4,activation='relu')(hidden_2)
```

```python
# here we combine both 
combine = concatenate([hidden_1,hidden_2_1])
```

```python
# applied the fully conneted layer and then regression on the combined layers
z = Dense(2,activation='relu')(combine)
z1 = Dense(1,activation='linear',)(z)
```

```python
# our model accept the braches of the 2 braches but output of the single valu 
model= Model(inputs=[input_1,input_2],outputs=z1)
```

```python
model.summary()
```
