```python
from keras.models import Model
```

```python
from keras.layers import *
```

```python
x = Input(shape=(3,))
```

```python
hidden_1 = Dense(128,activation='relu')(x)
hidden_2 = Dense(64,activation='relu')(hidden_1)

output_1 = Dense(1,activation='linear',name='linear')(hidden_2)
output_2 = Dense(1,activation='sigmoid',name='sigmoid')(hidden_2)
                                         
```

```python
model = Model(inputs=x ,outputs=[output_1,output_2])
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model,show_shapes=True, show_layer_names=True, expand_nested=True)
```
