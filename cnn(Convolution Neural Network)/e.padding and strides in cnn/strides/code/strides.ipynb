```python
from sklearn.datasets import fetch_openml

# Fetch the MNIST dataset from OpenML (may take a moment to download)
# setting as_frame=False returns data as numpy arrays instead of a pandas DataFrame
mnist = fetch_openml('mnist_784', version=1, as_frame=False)

# Extract features and targets
X, y = mnist.data, mnist.target

print(X.shape)  # Output: (70000, 784) -> 70,000 images, each 28x28 pixels flattened
print(y.shape)  # Output: (70000,) -> Labels '0' through '9'
```

```python
import tensorflow
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Conv2D,Flatten 
```

# Stride(usually not recommended)
**it tells us how many pixels the filter moves at each step while performing convolution**

```python
model = Sequential()
model.add(Conv2D(32,kernel_size=(3,3),strides=(2,2),activation='relu',input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),strides=(2,2),activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),strides=(2,2),activation='relu'))

model.add(Flatten()) # change in to 1D

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
model.summary()
```

⭐**The formula of the feature map with stride is [n-f/s]+1 where s = stride and it usually not recommend beacuse it loosed too much infomation of the image as u see in the model.summay**
