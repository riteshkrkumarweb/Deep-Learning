```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data() # mnist data from the keras
```

```python
X_train.shape # This is 3d array
```

```python
X_train[0] # this is the numpy array and the size of it is 28 by 28
```

```python
X_train[1].shape
```

```python
import matplotlib.pyplot as plt
plt.imshow(X_train[0]) # This is used to show the image of the numpy array
```

```python
y_train
```

```python
X_train = X_train/255
X_test = X_test/255 # here we do beacuse we put the all numpy in the range of 0 to 1
# and without it is in the range of 0 to 255 for the neural network value have to similar for best results ❤️
```

```python
X_train[0]
```

# Here we are going to make layers

```python
model = Sequential()
model.add(Flatten(input_shape=(28,28))) # Flatten make the higherdimenson data into 1d form from 28x28 to 784 input layers
model.add(Dense(128,activation='relu')) # u donot have to give the input layer it automatically through the flatten function
model.add(Dense(40,activation='relu'))
model.add(Dense(10,activation='softmax')) # output layer 10 beacuse output it multiclass and 0-9 output
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam',loss='sparse_categorical_crossentropy',metrics=['accuracy']) # which loss function we use # sparse automatically onehotencode the label
```

```python
history = model.fit(X_train,y_train,epochs=25,validation_split=0.2)
```

```python
y_prob = model.predict(X_test) # it find the probability that what is probability of
# of 0 is 7.63, 1 is 2.17,3 is 2.62 and so on see after the array
```

```python
y_prob
```

```python
y_predic = y_prob.argmax(axis=1) # argmax used maximum value in the index . 
# here we see that on which index we have the maximux value like in the fi
# in the first we see that 9.9999 on 7th index so probability of 7 is 9.999 .
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predic)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Accuracy vs val_accuracy')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('loss vs val_loss')
```
