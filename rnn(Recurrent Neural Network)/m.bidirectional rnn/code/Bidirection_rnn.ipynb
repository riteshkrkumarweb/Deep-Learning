```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import imdb
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, Bidirectional, Dense,SimpleRNN
```

```python
# Load the IMDb dataset
(x_train, y_train), (x_test, y_test)= imdb.load_data(num_words=10000)

# Pad sequences to have the same length
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

# Bidirectional Rnn

```python

from keras import Sequential, Input
from keras.layers import Embedding, Bidirectional, SimpleRNN, Dense

# Create a Sequential model (layers are added one after another)
model = Sequential()

# Define the input sequence: each sample contains 100 token IDs
model.add(Input(shape=(100,)))

# Convert each token ID into a dense vector of 32 numbers
# Input:  (batch_size, 100)
# Output: (batch_size, 100, 32)
model.add(Embedding(input_dim=10000, output_dim=32))

# First Bidirectional RNN layer
# 5 units in each direction: forward and backward
# return_sequences=True returns an output for every time step,
# allowing the next RNN layer to process the entire sequence
# Output shape: (batch_size, 100, 10)

model.add(Bidirectional(SimpleRNN(units=5, return_sequences=True)))# Bidirectional


# Second Bidirectional RNN layer (stacked on top of the first)
# Processes the sequence representations from the previous layer
# return_sequences=False returns the final combined representation
# Output shape: (batch_size, 10)
model.add(Bidirectional(SimpleRNN(units=5, return_sequences=False)))

# Binary classification output
# Sigmoid produces a probability between 0 and 1
model.add(Dense(units=1, activation="sigmoid"))

model.summary()
```

```python

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```
