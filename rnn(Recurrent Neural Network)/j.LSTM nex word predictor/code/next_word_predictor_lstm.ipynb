```python
for sentence in raw_text.split('\n'): # here we have split, where the new line is comming
  print(tokenizer.texts_to_sequences([sentence])) # after training vocabulary we converted it into sequence
```

#Create Input-Output pair with text

**The index is accessed from the outer list to the inner list — left to right.**

```python
input_sequences = []

for sentence in raw_text.split('\n'):
  tokenized_sentence = (tokenizer.texts_to_sequences([sentence])[0]) # to

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
for i in input_sequences:
  print(i)
```

**Here we impliment the zero padding so that the all the input is same but for the first we have find the maximum number of input**

```python
max_length = max([len(x) for x in input_sequences]) # we added the [] beacuse all the number in the lists , and get the maximum value
print(max_length)
```

**we apply zero padding to the input_sequence so that all input is equal , input cannot be different**

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
padded_sequences = pad_sequences(input_sequences, padding='pre') # it automatically do all things maximum padding
#per means previous and it is default without writing also
print(padded_sequences)
```

```python
print("Maximum padded length:", padded_sequences.shape)
print("Maximum padded length:", padded_sequences.shape[1]) # it fetcch the column we the same input and also the highest
```

**So here we determine the X and Y**

**Input Label**

```python
x = padded_sequences[:,:-1] # All the row and only remove the last column beacuse it is label input
print(x)
```

```python
x.shape # input_dimesion is 135 beacuse a row have 135 different input of same type
```

**Target Label/Output Label**

```python
y = padded_sequences[:,-1] # This is target label
print(y)
```

**So we take this data as a classification dataset and in the classification it is multiclass classification**

# Here we create the model

```python
import tensorflow
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Embedding, LSTM ,Input
```

**Emedding**output_dim=3 before training. You could instead choose output_dim=2 or output_dim=10.
That decision determines the shape of the embedding matrix. The actual values inside the matrix are learned by the model.

Example = (AI) token_id[3]

output_dim=3[0.2,0.3,0.07], output_dim of token_id(3)

ouput_dim=5[0.4,0.2,0,6,0.7,0.1] , output dim of token_di(3)

input_dim: Integer. Size of the vocabulary, i.e. maximum integer index + 1.

```python
model = Sequential()
model.add(Input(shape=(135,))) # input_shape each value per row
#vocablary is 429 but we write 430 beacuse word_embedding is started with zero index so 429+1extra = 430
# in embedding(vocublary_size,) vocularysize + 1 in the input_dim beacuse embedding do we the zero
model.add(Embedding(input_dim=430,output_dim=100))# we use embedding not direct token_ids , to caputure sementic meaning of words (like real word with context)
model.add(LSTM(units=150,))# We apply 150 nodes
model.add(Dense(units=430,activation='softmax'))# node is 430 and softmax it is multiclass classification
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam', metrics=['accuracy'])# for the multiple output we use categorical_cross entropy
```

```python
model.summary()
```

```python
model.fit(x,y,epochs=100)
```

#Here we predict the text

```python
import numpy as np

text = "lethal"


# first tokenize
token_text = tokenizer.texts_to_sequences([text]) [0] # added[0]we donot want the nested list
# padding
padded_token_text = pad_sequences([token_text])
# predict
postion = np.argmax(model.predict(padded_token_text)) #np.argmax() returns the index (position) of the largest value in an array.
print(postion)

for word,index in tokenizer.word_index.items():
  if index == postion:
    print("The predicted word is: ",word)
```
