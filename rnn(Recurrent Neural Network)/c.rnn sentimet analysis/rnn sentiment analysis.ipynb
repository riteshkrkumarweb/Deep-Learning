#Tokenization and Integer Encoding

```python
import keras
from keras.layers import TextVectorization # tokenizer only thing is to break the text into smaller piece that it
#it donot change into the integer here and ther is combined menthod with it ouput_mode = int it it integer encoding

tokenizer = TextVectorization(output_mode='int')
```

```python
tokenizer.adapt(docs) # like tokenizer.fit_on_text(docs) with the old tuturial but removed
# it also do automatically padding so u saw many zeroes in tokenizer(docs)output
```

```python
word_after_tokenizer= tokenizer.get_vocabulary()
print(word_after_tokenizer)
```

```python
sequence = tokenizer(docs)
print(sequence)
```

#Embedding

```python
from keras import Sequential
from keras.layers import Embedding,Dense,Input
```

output_dim=4(hyperpramter) 2D/3D/4D is simply the number of coordinates used to locate each token in the embedding space. The larger the dimension, the more numerical capacity the representation has, although higher dimension does not automatically mean a better embedding.

Input(shape=9) means highest sencentence word

1st u have to write the unique vocabulary like 40 unique word get by tokenizer.

```python
model = Sequential()
model.add(Input(shape=(9,)))
model.add(Embedding(40,output_dim=4))
#ouput_dim each token gets a vector containing 4 numbers
#output_dim determines the dimensionality of the embedding space. Similarity between tokens can then be measured using their vectors in that space.
```

```python
model.summary()
```

```python
model.compile('adam','accuracy')
```

# This is Embedding first we do the integer Encoding then Embedding

 when we have set the output dim 4 then we get the 4 dimesion for each word [ 0.03391692  0.0376677  -0.03837267 -0.02630422] like this

```python
pred = model.predict(sequence)
print(pred)
```

# Lets try on imbd dataset

```python
from keras import Sequential
from keras.layers import Flatten,Embedding,SimpleRNN,Dense
```

```python
from os import X_OK
from keras.datasets import imdb
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data()
```

Before padding → different sequence lengths → no fixed second dimension.so not columns given

```python
x_train.shape
```

The review has all different sentence so we do the padding

```python
from keras.utils import pad_sequences
x_train = pad_sequences(x_train,padding='post')
x_test = pad_sequences(x_test,padding='post')
```

```python
x_train.shape # each row contained 2494 items
```

```python
print(x_train.max()) # maximun totol number of ids create  with the integer encoding
```

#Embedding

10000 means 100 unique tokens in the vocabulary (more precisely, 10000 token IDs the layer can look up).

Vocabulary
──────────────

0 → PADDING

1 → UNKNOWN Words

2 → I

3 → Love

4 → apple

...

99 → banana


2494 = 2494 numbers for each token's embedding vector(each embeddding vector item)

```python
model = Sequential()

model.add(Input(shape=(2494,))) #input shape
model.add(Embedding(input_dim = 88587,output_dim=2))#ouput_dim(hyperparameter)
#88587 vocabulary size , unique token and 2d dimension space totol number of ids with the integer encoding
#i donot know the exactly unique so i set it to 10000

model.add(SimpleRNN(32,return_sequences=False))# loop traning of 32

model.add(Flatten())
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```
