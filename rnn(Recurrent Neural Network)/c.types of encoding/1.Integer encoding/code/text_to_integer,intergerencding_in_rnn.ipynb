⭐**Tokenizer a tool that breaks raw text into smaller pieces called tokens, like words, parts of words, or characters, and translates them into numbers that a computer can understand**

 **"AI is fun!"  Tokens: ["AI", "is", "fun", "!"]**


```python
import keras
from keras.layers import TextVectorization # tokenizer only thing is to break the text into smaller piece that it
#it donot change into the integer here and ther is combined menthod with it ouput_mode = int it it integer encoding

tokenizer = TextVectorization(output_mode='int')
```

```python
tokenizer.adapt(docs) # like tokenizer.fit_on_text(docs) with the old tuturial but removed
```

```python
word_after_tokenizer= tokenizer.get_vocabulary()
print(word_after_tokenizer)
```

```python
vocab = tokenizer.get_vocabulary() #[UNK]in output: Unknown It is a special token used when the model encounters a word that was not present in the vocabulary during adapt().
                                    # index 0 is reserved for padding so that vector of text is same
for index, word in enumerate(word_after_tokenizer):
    print(index, "→", word)
```

#Lets now work on new dataset like imdb review on movie ,means sentiments training

```python
from keras.datasets import imdb
from keras.layers import Dense,SimpleRNN,Flatten
from keras import Sequential
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data()
```

```python
from keras.utils import  pad_sequences

x_train = pad_sequences(x_train,padding='post')# maxlen=4  sets the maximum/fixed sequence length to 4. if u have 5 word the
                                                          # then one word is removed so set the maximun so review word cannot delete also
                                                          #set max = 5000 also make the sentence add zero padding .
x_test = pad_sequences(x_test,padding='post')# max_len basically means trimming the word
```

```python
len(x_train[1])
```

```python
len(x_train[2])
```

```python
model = Sequential()
model.add(SimpleRNN(32,input_shape=(2494,1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```
