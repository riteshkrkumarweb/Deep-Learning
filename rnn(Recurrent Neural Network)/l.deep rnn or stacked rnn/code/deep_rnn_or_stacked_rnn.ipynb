```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import imdb
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, SimpleRNN, Dense
```

**The IMDB dataset's word index contains approximately 88,000 unique word entries.**

```python
# Load the IMDb dataset
(x_train, y_train), (x_test, y_test)= imdb.load_data(num_words=10000)

# Pad sequences to have the same length
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

**Total Vocabulary Size is 88000**

**return_sequences=True: RNN return the output for every time step**

**When stacking two RNN layers, the first layer usually needs return_sequences=True so the second layer receives the full sequence.**

**If you have 3 RNN layers, normally you set return_sequences=True for the first and second layers, and False or Leave default is False for the third (final) layer when you need one prediction per sequence.**

```python
from keras import Input
# Define the RNN model
model = Sequential()
model.add(Input(shape=(100,)))
model.add(Embedding(10000, 32)), # Embedding layer to convert words to vectors
model.add(SimpleRNN(5, return_sequences=True)),#
model.add(SimpleRNN(5)),#one more stacked rnn
model.add(Dense(1, activation='sigmoid'))

# RNN layer with 5 units
# Another RNN layer with 5 units
# Output layer for binary classification

model.summary()
```

```python
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```
