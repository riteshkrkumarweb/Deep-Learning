```python
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split

X, y = make_circles(n_samples=100, factor=0.1, noise=0.3,)
X_train, X_test, y_train, y_test = train_test_split(X, y,)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X[:,0],X[:,1],c=y)
plt.title('Dataset')
```

```python
import tensorflow 
from tensorflow.keras.layers import Dense 
from tensorflow.keras import Sequential
```

```python
model = Sequential()

model.add(Dense(16, activation='relu', input_dim=2))
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

# With the Normal SGD 

```python
# With the Normal SGD 
from keras.optimizers import SGD 
sgd = SGD(learning_rate=0.01, momentum=0.0, nesterov=False,)

model.compile(loss='binary_crossentropy',metrics=['accuracy'],optimizer=sgd)
```

```python
model.fit(X_train,y_train,validation_split=0.2,epochs=100)
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt
plt.title('Normal SGD ')
# Plotting decision regions
plot_decision_regions(X, y, clf=model, legend=2)
```

```python
modelwith_normal_SgD = model.predict(X_test)
y_predwithnormal_sgd = ( modelwith_normal_SgD >= 0.5 ).astype('int')
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predwithnormal_sgd)
```

# Accuracy is 76 % | with the normal SGD on test Data

# SGD with Momentum 

```python
modelwithmomentum = Sequential()

modelwithmomentum.add(Dense(16, activation='relu', input_dim=2))
modelwithmomentum.add(Dense(8, activation='relu'))
modelwithmomentum.add(Dense(1, activation='sigmoid'))
```

```python
# SGD with Momentum 
#momentum = Beta 
# mometum = 0.9 - remember more previous momentum 
# mometum = 0.5 - remember less previous momentum 
# momntum  = 0 equivalent to the normal SGD 
from keras.optimizers import SGD 
sgdwithmomentum = SGD(learning_rate=0.01, momentum=0.9, nesterov=False,)

modelwithmomentum.compile(loss='binary_crossentropy',metrics=['accuracy'],optimizer=sgdwithmomentum)
```

```python
modelwithmomentum.fit(X_train,y_train,validation_split=0.2,epochs=100)
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt
# Plotting decision regions
plt.title('SGD with Momentum')
plot_decision_regions(X, y, clf=modelwithmomentum, legend=2)
```

```python
model_sgdwith_momentum = modelwithmomentum.predict(X_test)
y_predsgdwithmomentum = ( model_sgdwith_momentum >= 0.5 ).astype('int')
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predsgdwithmomentum)
```

# Accuracy is 88% |  SGD with Momentum on test Data 

# NAG optimizer (Nestrov Accelareted Gradient) Improved Version of the Momentum

```python
modelwith_nag_optimizer = Sequential()

modelwith_nag_optimizer.add(Dense(16, activation='relu', input_dim=2))
modelwith_nag_optimizer.add(Dense(8, activation='relu'))
modelwith_nag_optimizer.add(Dense(1, activation='sigmoid'))
```

```python
# NAG (nestrov Acceleration Grdient )  nestrov = True
#momentum = Beta 
# mometum = 0.9 - remember more previous momentum 
# mometum = 0.5 - remember less previous momentum 
# momntum  = 0 equivalent to the normal SGD 
from keras.optimizers import SGD 
nagoptimizer = SGD(learning_rate=0.01, momentum=0.9, nesterov=True,)
modelwith_nag_optimizer.compile(loss='binary_crossentropy',metrics=['accuracy'],optimizer=nagoptimizer)
```

```python
modelwith_nag_optimizer.fit(X_train,y_train,validation_split=0.2,epochs=100)
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt
plt.title('NAG Optimizer')
# Plotting decision regions
plot_decision_regions(X, y, clf=modelwith_nag_optimizer, legend=2)
```

```python
modelwithNAG = modelwith_nag_optimizer.predict(X_test)
y_predwitNAG = ( modelwithNAG >= 0.5 ).astype('int')
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predwitNAG)
```

# Accuracy With NAG is 92 % on test Data
